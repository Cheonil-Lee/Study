```python
import cv2
import os
from ultralytics import YOLO

# 모델 로드
model = YOLO('/content/drive/MyDrive/Study/redzone/train2/weights/best.pt')  # 사전 훈련된 모델 로드

# 동영상 파일 경로
video_path = "/content/drive/MyDrive/Study/redzone/250115_video/camera2_20250115_153101.avi"  # 비디오 파일 경로

# 비디오 파일 이름 추출 및 결과 저장 경로 설정
video_name = os.path.basename(video_path)  # 파일 이름 추출
video_name_without_ext = os.path.splitext(video_name)[0]  # 확장자 제거
#output_path = f"/content/drive/MyDrive/Study/redzone/result_0115/{video_name_without_ext}_camera1_result.avi"  # 결과 저장 경로

# 비디오 캡처 객체 생성
cap = cv2.VideoCapture(video_path)

# 비디오 작성을 위한 설정
fourcc = cv2.VideoWriter_fourcc(*'XVID')
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

while True:
    ret, frame = cap.read()
    if not ret:
        break  # 비디오의 끝에 도달

    # 객체 탐지 수행
    results = model.predict(frame, save=False)

    # 결과 시각화
    img = results[0].plot(line_width=2, font_size=10, pil=True)

    # 결과 프레임 저장 (변환 없이 저장)
    out.write(img)  # img가 BGR 형식일 경우 변환하지 않음

# 자원 해제
cap.release()
out.release()
cv2.destroyAllWindows()
```
